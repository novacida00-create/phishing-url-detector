# 2. Preprocessing

Satu baris dengan label kosong dibuang. URL duplikat juga dibuang supaya tidak muncul di dua split.
Pembagian data nanti dilakukan berdasarkan domain, jadi URL dari domain yang sama tidak boleh
tersebar ke data latih dan data uji.

```python
import sys
from pathlib import Path

ROOT = Path.cwd()
if not (ROOT / "app").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

import pandas as pd
from train_model import domain_group

RAW = ROOT / "dataset" / "raw" / "LegitPhish Dataset" / "url_features_extracted1.csv"
PROCESSED = ROOT / "dataset" / "processed"
PROCESSED.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(RAW)
bersih = df.dropna(subset=["ClassLabel"]).drop_duplicates(subset=["URL"]).reset_index(drop=True)
bersih["ClassLabel"] = bersih["ClassLabel"].astype(int)

print("Sebelum:", len(df), "sesudah:", len(bersih))
print(bersih["ClassLabel"].value_counts().to_string())
```

```python
bersih["domain_group"] = bersih["URL"].map(domain_group)
print("Jumlah grup domain:", bersih["domain_group"].nunique())
print(bersih["domain_group"].value_counts().head(10).to_string())

bersih.to_csv(PROCESSED / "data_bersih.csv", index=False)
print("\nTersimpan di", PROCESSED / "data_bersih.csv")
```
