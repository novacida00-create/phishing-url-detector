# 1. Eksplorasi dataset

File CSV diambil dari LegitPhish Dataset versi 2 (Mendeley Data, DOI 10.17632/hx4m73v2sf.2).
Tahap ini hanya melihat bentuk data dan memeriksa catatan jumlah URL di halaman dataset.

```python
import sys
from pathlib import Path

ROOT = Path.cwd()
if not (ROOT / "app").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

import pandas as pd

RAW = ROOT / "dataset" / "raw" / "LegitPhish Dataset" / "url_features_extracted1.csv"
df = pd.read_csv(RAW)

print("Bentuk data:", df.shape)
print("Kolom:", df.columns.tolist())
```

```python
print(df["ClassLabel"].value_counts(dropna=False))
print("\nBaris dengan label kosong:")
print(df[df["ClassLabel"].isna()][["URL", "ClassLabel"]].to_string(index=False))
```

```python
print("URL duplikat:", df["URL"].duplicated().sum())
print("\nJumlah nilai kosong tiap kolom:")
print(df.isna().sum().to_string())
```
